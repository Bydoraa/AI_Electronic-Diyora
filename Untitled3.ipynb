```python
# Required libraries
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt

# 1. Create data
data = {
    'Income': [60000, 55000, 70000, 50000, 52000,
               48000, 45000, 40000, 30000, 35000,
               60000, 52000, 70000, 48000, 45000,
               50000, 51000, 49000, 53000, 55000],
    'Age':    [32, 35, 40, 30, 33,
               28, 25, 22, 27, 20,
               25, 28, 29, 30, 31,
               25, 29, 35, 31, 30],
    'Buy':    ['Yes', 'Yes', 'Yes', 'Yes', 'Yes',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'Yes', 'Yes']
}

# Create DataFrame
df = pd.DataFrame(data)

# 2. Print DataFrame
print("--- DataFrame Output ---")
print(df)
```

```python
# Scatter plot
plt.figure(figsize=(8,6))

# Buy = Yes
yes = df[df['Buy'] == 'Yes']
plt.scatter(yes['Income'], yes['Age'], color='green', label='Buy: Yes', s=100, marker='o')

# Buy = No
no = df[df['Buy'] == 'No']
plt.scatter(no['Income'], no['Age'], color='red', label='Buy: No', s=100, marker='x')

# 그래프 설정
plt.xlabel('Income')
plt.ylabel('Age')
plt.title('Income vs Age Scatter Plot (Buy Yes/No)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 3. Define Features (X) and Target (y)
X = df[['Income', 'Age']]
y = df['Buy']

# 4. Train Decision Tree Model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# 5. Visualize the Decision Tree
plt.figure(figsize=(12, 8))
tree.plot_tree(model,
               feature_names=['Income', 'Age'],
               class_names=['No', 'Yes'],
               filled=True)
plt.title("Decision Tree for 'Buy' Prediction")
plt.show()
```

```python
# 5️. Prediction
sample = pd.DataFrame({'Income':[ 47000, 51000], 'Age':[29, 31]})
pred = model.predict(sample)
print(pred)  # ['Yes', 'No']

####  Part 5 ###
# 6. Model rule print. not function f(x)
from sklearn.tree import export_text

# 트리의 규칙을 텍스트 형태로 출력
tree_rules = export_text(model, feature_names=['Income', 'Age'])
print(tree_rules)
```

```python
# Decision boundary
import numpy as np
import matplotlib.pyplot as plt

# 4. Scatter plot and decision boundary visualization
plt.figure(figsize=(10, 6))

# Create meshgrid for decision boundary
x_min, x_max = X['Income'].min() - 5000, X['Income'].max() + 5000
y_min, y_max = X['Age'].min() - 2, X['Age'].max() + 2

xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 500),
    np.arange(y_min, y_max, 0.5)
)

# Predict for each point in the meshgrid
grid = np.c_[xx.ravel(), yy.ravel()]

# Use DataFrame to preserve feature names
grid_df = pd.DataFrame(grid, columns=['Income', 'Age'])

Z = model.predict(grid_df)

# Convert predictions into numerical values
Z = np.where(Z == 'Yes', 1, 0)
Z = Z.reshape(xx.shape)

# Plot decision boundary
plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')

# Plot original data points
plt.scatter(
    X['Income'],
    X['Age'],
    c=np.where(y == 'Yes', 1, 0),
    cmap='coolwarm',
    edgecolors='black'
)

plt.xlabel('Income')
plt.ylabel('Age')
plt.title('Decision Boundary - Decision Tree Classifier')
plt.colorbar(label='Predicted Class')
plt.show()
```

```python
# New customer data
new_income = 62000
new_age = 32

# Create DataFrame
new_data = pd.DataFrame({
    'Income': [new_income],
    'Age': [new_age]
})

# Predict new customer's class
prediction = model.predict(new_data)

# Plot decision boundary
plt.figure(figsize=(10, 6))

plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')

# Plot original data
plt.scatter(
    X['Income'],
    X['Age'],
    c=np.where(y == 'Yes', 1, 0),
    cmap='coolwarm',
    edgecolors='black',
    label='Existing Customers'
)

# Plot new customer as a red star
plt.scatter(
    new_income,
    new_age,
    color='yellow',
    marker='*',
    s=300,
    edgecolors='black',
    label='New Customer'
)

# Add annotation
plt.annotate(
    f'New Customer: {prediction[0]}',
    (new_income, new_age),
    xytext=(10, 15),
    textcoords='offset points',
    fontsize=11,
    fontweight='bold'
)

plt.xlabel('Income')
plt.ylabel('Age')
plt.title('Decision Boundary with New Customer Diyora')
plt.legend()
plt.colorbar(label='Predicted Class')
plt.show()

# Display prediction
print("New Customer Prediction:", prediction[0])
```
